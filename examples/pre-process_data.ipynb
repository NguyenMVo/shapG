```python
from sklearn.preprocessing import LabelEncoder


# Creating a LabelEncoder object
label_encoder = LabelEncoder()

# Columns to apply label encoding on (only for binary or ordinal categories)
columns_to_label_encode = ['family_history','cancer_stage', 'gender',  'smoking_status','treatment_type']

# Applying label encoding to each column
for col in columns_to_label_encode:
    df1[col] = label_encoder.fit_transform(df1[col])

# Applying one-hot encoding for 'treatment_type'
#df1 = pd.get_dummies(df1, columns=['treatment_type'], drop_first=True)

# Display the updated dataframe
print(df.head())
```

```python
# Display the updated dataframe
print(df1.head())
```

```python
# Dropping the 'id' and 'country' columns
df1 = df1.drop(columns=['id', 'country'])

# Display the updated dataframe
df1.head()
```

```python
#shape define the columns and rows of dataset
#18 Columns & 325000 rows
df1.shape
```

```python
# Ensure both 'diagnosis_date' and 'beginning_of_treatment_date' are in datetime format
df1['diagnosis_date'] = pd.to_datetime(df1['diagnosis_date'])
df1['beginning_of_treatment_date'] = pd.to_datetime(df1['beginning_of_treatment_date'])

# Calculate the treatment delay (time between diagnosis and start of treatment)
df1['treatment_delay_days'] = (df1['beginning_of_treatment_date'] - df1['diagnosis_date']).dt.days

# Drop the original date columns if they're no longer needed
df1 = df1.drop(columns=['diagnosis_date'])

# Display the updated dataframe
print(df1.head())
```

```python
# Ensure the columns are in datetime format
df1['beginning_of_treatment_date'] = pd.to_datetime(df1['beginning_of_treatment_date'])
df1['end_treatment_date'] = pd.to_datetime(df1['end_treatment_date'])

# Calculate the duration of treatment in days
df1['treatment_duration_days'] = (df1['end_treatment_date'] - df1['beginning_of_treatment_date']).dt.days

# Drop the original date columns
df1 = df1.drop(columns=['beginning_of_treatment_date', 'end_treatment_date'])

# Display the updated dataframe
print(df1.head())
```
